### Handling outlier: We can find one common value and replace the outlier(Both positive and negative skewed values). This is called "CAPPING"


```python
quantity_upper_limit = df["Quantity"].quantile(0.99) #q3bvalue
quantity_upper_limit 
```

```python
quantity_lower_limit =  df["Quantity"].quantile(0.01)
quantity_lower_limit 
```

```python
IQR= quantity_upper_limit - quantity_lower_limit
upper_fence = quantity_upper_limit + (IQR*1.5)
lower_fence = quantity_lower_limit - (IQR*1.5)
upper_fence
```

```python
lower_fence
```

#### Capping

```python
df["Quantity"] = np.where(df["Quantity"]> quantity_upper_limit,upper_fence,
         np.where(df["Quantity"]< quantity_lower_limit,lower_fence,df["Quantity"]))
```

```python
df[df["Quantity"]<lower_fence]
```

```python
df[df["Quantity"]>upper_fence]
```

#### Function to analyse product sales:
#### Question: Create a function to analyze sales perfromance of a specific product accross all countries.

```python
def analyze_product_sales(product_description,df):


    # filter data for the product
    product_data=df[df["Description"]==product_description]
     # calculate total_sales amount and the quantity sold:
    total_sales=product_data["Sales"].sum()
    total_quantity=product_data["Quantity"].sum()

    # Calculate sales performance accross all countries:
    country_perfromance=product_data.groupby("Country").agg({"Sales":"sum","Quantity":"sum"})
    country_perfromance=country_perfromance.reset_index()


    return {
    "Total Sales Amount" : total_sales,
    "Total Quantity Sold" :total_quantity,
    "Sales performance by country": country_perfromance
    }


# example usage

product_performance=analyze_product_sales("WHITE HANGING HEART T-LIGHT HOLDER",df)
```
